```python
import requests
from bs4 import BeautifulSoup
from urllib.parse import urljoin
```

```python
# Target URL
base_url = "https://itrafik.mbjb.gov.my/itrafik/"

# Make request to the page
response = requests.get(base_url)
soup = BeautifulSoup(response.content, "html.parser")

# Find all <img> tags
img_tags = soup.find_all("img")

# Extract src from each <img> and join it with the base URL
image_links = [urljoin(base_url, img['src']) for img in img_tags if 'src' in img.attrs]

# Print or save the links
for link in image_links:
    print(link)
```

```python
import requests
import os
from urllib.parse import urlparse
import re

# List of URLs
urls = image_links

# Directory to save the images
save_dir = "mbjb_traffic_images"
os.makedirs(save_dir, exist_ok=True)

for url in urls:
    if "/cctv_snapshots/" not in url:
        continue  # Skip logos, QR codes, etc.

    try:
        # Extract location, date, time from URL
        match = re.search(r'/cctv_snapshots/(L\d+_Cam\d+)/(\d{4}-\d{2}-\d{2})/pic_001/([\d.]+)\[', url)
        if not match:
            print(f"Skipped (No match): {url}")
            continue

        location, date, time_str = match.groups()
        time_str = time_str.replace('.', '-')  # Clean timestamp

        # Create a filename like L01_Cam1_2025-04-13_16-49-23.jpg
        filename = f"{location}_{date}_{time_str}.jpg"
        filepath = os.path.join(save_dir, filename)

        # Download the image
        response = requests.get(url)
        if response.status_code == 200:
            with open(filepath, 'wb') as f:
                f.write(response.content)
            print(f"Saved: {filename}")
        else:
            print(f"Failed to download: {url}")
        print("Finish saving")
    except Exception as e:
        print(f"Error processing {url}: {e}")
```
